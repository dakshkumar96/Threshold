## Load

```python
from pathlib import Path
import pandas as pd

cwd = Path.cwd()
if (cwd / "data" / "raw").exists():
    project_root = cwd
elif (cwd.parent / "data" / "raw").exists():
    project_root = cwd.parent
else:
    raise FileNotFoundError("Could not find data/raw/")

csv_path = project_root / "data" / "raw" / "sponsor_register_2026-07-28.csv"
df = pd.read_csv(csv_path)
print(len(df), "rows")
df.head()
```

## Columns

```python
for col in df.columns:
    print("=" * 50)
    print("name:", col)
    print("dtype:", df[col].dtype)
    print("nulls:", int(df[col].isna().sum()))
    print("unique:", int(df[col].nunique()))
    print("examples:", df[col].dropna().unique()[:10].tolist())
```

## Four questions

### Q1 — Company ID?

```python
df.columns.tolist()
```

**Names only** — no company number column.

### Q2 — Skilled Worker route?

```python
print(df["Route"].value_counts())

skilled = df[df["Route"] == "Skilled Worker"]
print("\nSkilled Worker rows:", len(skilled))
```

**Yes** — filter `Route == "Skilled Worker"`.

### Q3 — Licence rating?

```python
print(df["Type & Rating"].value_counts().head(20))

a_count = df["Type & Rating"].str.contains("A rating", na=False).sum()
b_count = df["Type & Rating"].str.contains("B rating", na=False).sum()
print(f"\nRows with A rating: {a_count}")
print(f"Rows with B rating: {b_count}")
```

**Yes** — in `Type & Rating` (A/B).

### Q4 — Region?

```python
print("Town/City nulls:", int(df["Town/City"].isna().sum()))
print("County nulls:", int(df["County"].isna().sum()))
print("\nTop Town/City values:")
print(df["Town/City"].value_counts(dropna=False).head(10))
print("\nTop County values:")
print(df["County"].value_counts(dropna=False).head(10))
```

**Yes** — `Town/City` + `County` (County has many blanks).
