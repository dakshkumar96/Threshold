**Question:** What is linked to sponsors staying on the register (2023-09 → 2026-07)?

```python
from pathlib import Path
import pandas as pd

root = Path.cwd() if (Path.cwd() / "data" / "processed").exists() else Path.cwd().parent
surv = pd.read_parquet(root / "data" / "processed" / "survival_table.parquet")

print("Companies:", len(surv), "Events:", int(surv["event"].sum()))
display(
    surv.groupby("region").agg(
        n=("event", "size"),
        exit_rate=("event", "mean"),
        median_days=("duration_days", "median"),
    ).sort_values("exit_rate", ascending=False)
)
```

## Finding

London sponsors show a higher raw exit rate (**~10.1%**) than Scotland (**~8.3%**). Region log-rank p ≈ **7e-5**. Cox HR for Scotland vs London reference ≈ **0.84**.

Rating is mostly “A”; it is a weak discriminator in this sample.

## Limitations

Irregular snapshots, names-only IDs, no sector, simplified censoring — see `ANALYSIS.md`.
