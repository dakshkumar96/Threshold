# Phase 2 — Jobs + sponsor match

Fetches UK jobs for a **role you choose** (Reed + Adzuna), then fuzzy-matches employers to Skilled Worker sponsors.

Change `ROLE` below and re-run.

```python
import sys
from pathlib import Path

cwd = Path.cwd()
project_root = cwd if (cwd / "data" / "processed").exists() else cwd.parent
sys.path.insert(0, str(project_root / "src"))

from run_jobs_pipeline import fetch_all_jobs, run
from match_sponsors import match_jobs_to_sponsors

ROLE = "data analyst"  # change this to any role, e.g. "software engineer"
ROLE
```

```python
jobs = fetch_all_jobs(ROLE, max_per_source=250)
print("Total rows:", len(jobs))
print(jobs["source"].value_counts())
jobs[["source", "title", "company_raw", "location"]].head(10)
```

```python
matched = match_jobs_to_sponsors(jobs)
n = len(matched)
n_sponsor = int(matched["is_sponsor"].sum())
print(f"Matched to sponsor: {n_sponsor} / {n} ({100 * n_sponsor / n:.1f}%)")

matched.loc[
    matched["is_sponsor"],
    ["title", "company_raw", "matched_company_key", "match_score", "still_active", "location"],
].sort_values("match_score", ascending=False).head(20)
```

```python
jobs_path, matched_path = run(ROLE)
jobs_path.name, matched_path.name
```
